# Will create a 2d array contains 3 columns with assignment,mid term, end term marks, Each columns has weights to calculate, the we have to calculate the weighted marks.


```python
marks = np.array([[23,24,45],[22,25,44],[21,23,46]])
weights = np.array([0.2,0.3,0.4])
finalMarks = marks @ weights
print(finalMarks)
ans = np.dot(marks,weights)
ans = np.dot(weights,marks)
print(ans)
ans1 = finalMarks.sum()
print(ans1)
```

```python
marks = np.array([[23,24,45],[22,25,44],[21,23,46]])
weights = np.array([0.2,0.3,0.4])
finalMarks = marks * weights
print(finalMarks)
ans = np.dot(marks,weights) #Dot Product(Matrix multiplication)
print(ans)
ans1 = finalMarks.sum()
print(ans1)
```
