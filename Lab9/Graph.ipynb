## Graphs

```python
import matplotlib.pyplot as plt
```

# 1) Line Graph

a) Single Line Graph


```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.plot(x, y, marker='o')
plt.title("Single Line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

b) Double Line Graph

```python
x = [1, 2, 3, 4, 5]
y1 = [10, 20, 15, 25, 30]
y2 = [12, 17, 29, 30, 28]
plt.plot(x, y1, marker='s',label = "Line 1")
plt.plot(x, y2, marker='o',label = "Line 2")
plt.title("Double Line Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

## 2) Bar Graph

a) Single bar graph

```python
x = ["CSE", "ECE", "ME", "Civil"]
y = [80, 70, 60, 75]

plt.bar(x, y)
plt.title("Single Bar Graph")
plt.xlabel("Departments")
plt.ylabel("Marks")
plt.show()
```

b) Double bar graph

```python

x = [1, 2, 3, 4]
y1 = [80, 70, 60, 75]
y2 = [75, 85, 65, 80]

plt.bar([i - 0.2 for i in x], y1, width=0.4, label="Boys")
plt.bar([i + 0.2 for i in x], y2, width=0.4, label="Girls")
plt.title("Double Bar Graph")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()
plt.show()
```

## 3) Histogram

```python
data = [10, 20, 20, 30, 30, 30, 40, 40, 50, 60]

plt.hist(data, bins=5, edgecolor='black')
plt.title("Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

## 4) Pie Chart

```python
labels = ["CSE", "ECE", "ME", "Civil"]
sizes = [40, 25, 20, 15]

plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Department Distribution")
plt.show()
```
