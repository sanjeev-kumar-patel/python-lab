```python
## Creating a cursor
cursor = connection.cursor()
print(cursor)
```

```python
query = """Create table if not exists student(
    sId integer primary key,
    sName text,
    smarks real,
    class text
    )"""
```

```python
cursor.execute(query)
print("Table Created")
```

```python
connection.commit()
```

## Selecting particular column

```python
cursor.execute("Select sName,sMarks from student")
rows = cursor.fetchall()
for r in rows:
    print(r)
```

## Where clause

```python
query4 = "SELECT * FROM student WHERE class = 'CSE'"

cursor.execute(query4)

rows = cursor.fetchall()

for r in rows:
    print(r)
```

```python
query4 = "SELECT * FROM student WHERE class = ?"

cursor.execute(query4, ("CSE",))

rows = cursor.fetchall()

for r in rows:
    print(r)
```

## Update


```python
cursor.execute("update student set sMarks = ? where id = ?",(91,22))
connection.commit()
```
